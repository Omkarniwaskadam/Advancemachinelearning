### Apriori Algorithm Implementation

The Apriori algorithm is an influential algorithm for mining frequent itemsets and deriving association rules. It operates on the principle that if an itemset is frequent, then all of its subsets must also be frequent. This property helps to prune the search space efficiently.

**Steps:**
1.  **Generate Candidate Itemsets:** Create all possible itemsets of a certain length.
2.  **Prune:** Remove candidate itemsets that have an infrequent subset.
3.  **Count Support:** Count the occurrences of remaining itemsets in the dataset.
4.  **Filter by Minimum Support:** Keep only those itemsets whose support is greater than or equal to a predefined minimum support threshold.
5.  **Repeat:** Increment the itemset length and repeat until no more frequent itemsets can be found.

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# Sample Dataset (list of transactions)
transactions = [
    ['milk', 'bread', 'butter'],
    ['milk', 'diapers', 'beer'],
    ['bread', 'diapers', 'cola'],
    ['milk', 'bread', 'diapers', 'beer'],
    ['bread', 'butter', 'cola']
]

# Convert the list of transactions into a one-hot encoded DataFrame
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
df = pd.DataFrame(te_ary, columns=te.columns_)

print("One-Hot Encoded DataFrame:")
display(df.head())
```

```python
# 1. Find Frequent Itemsets using Apriori
# min_support is the minimum support threshold (e.g., 0.5 means at least 50% of transactions)

frequent_itemsets = apriori(df, min_support=0.4, use_colnames=True)

print("\nFrequent Itemsets (min_support=0.4):")
display(frequent_itemsets)
```

```python
# 2. Generate Association Rules from Frequent Itemsets
# min_threshold is the minimum confidence threshold (e.g., 0.7 means at least 70% confidence)

rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)

print("\nAssociation Rules (min_confidence=0.7):")
display(rules)

# You can also sort the rules by other metrics like 'lift'
print("\nAssociation Rules sorted by Lift:")
display(rules.sort_values(by='lift', ascending=False))
```
